```python
import os
import glob
import cv2
from pathlib import Path
from ultralytics import YOLOWorld
import shutil

# --- Ensure raw_onion_images is populated ---
# Clear existing raw_onion_images to ensure a clean re-extraction
if Path("./raw_onion_images").exists():
    shutil.rmtree("./raw_onion_images")

# Ensure the zip file is present by re-downloading it (if not already there or deleted)
# This assumes the kaggle.json is set up, as shown in previous cells.
!kaggle datasets download -d rahul14112003/onion-grading-dataset -p /content --force

# Re-unzip the dataset to ensure images are present, specifying full path
!unzip -q /content/onion-grading-dataset.zip -d ./raw_onion_images
print("Ensured raw_onion_images is populated.")
# --- End ensure raw_onion_images populated ---

# --- Start Image Preparation ---
# Search for all jpg/png files anywhere inside raw_onion_images
found_images = glob.glob("./raw_onion_images/**/*.*", recursive=True)
valid_images = [img for img in found_images if img.lower().endswith(('.jpg', '.jpeg', '.png'))]

print(f"Total raw images found: {len(valid_images)}")

target_dir = Path("./flat_onion_images")
# Clear existing flat images directory if it exists and recreate
if target_dir.exists():
    shutil.rmtree(target_dir)
target_dir.mkdir(parents=True, exist_ok=True)

for i, img_path in enumerate(valid_images):
    # Copy with a clean unique name to avoid subfolder collisions
    shutil.copy(img_path, target_dir / f"onion_{i}.jpg")

print(f"Successfully collected {len(list(target_dir.glob('*.jpg')))} images ready for labeling!")
# --- End Image Preparation ---

# 1. Setup YOLO dataset directories
dataset_root = Path("./onion_yolo_dataset")
train_img_dir = dataset_root / "train" / "images"
train_lbl_dir = dataset_root / "train" / "labels"
val_img_dir = dataset_root / "valid" / "images"
val_lbl_dir = dataset_root / "valid" / "labels"

for p in [train_img_dir, train_lbl_dir, val_img_dir, val_lbl_dir]:
    p.mkdir(parents=True, exist_ok=True)

# 2. Load open-vocabulary zero-shot detector
model = YOLOWorld("yolov8s-world.pt")

# Set the classes to find
classes = ["onion", "rotten onion", "sprouted onion", "damaged onion", "round coin"]
model.set_classes(classes)

# 3. Process ALL flattened images (changed from [:350] to all)
image_files = sorted(glob.glob("./flat_onion_images/*.jpg"))
print(f"Auto-labeling {len(image_files)} images using YOLO-World...")

for idx, img_path in enumerate(image_files):
    img_name = Path(img_path).name
    lbl_name = Path(img_path).stem + ".txt"

    # 80/20 train/valid split
    is_val = (idx % 5 == 0)
    dest_img = val_img_dir / img_name if is_val else train_img_dir / img_name
    dest_lbl = val_lbl_dir / lbl_name if is_val else train_lbl_dir / lbl_name

    img = cv2.imread(img_path)
    if img is None:
        continue
    h, w, _ = img.shape
    cv2.imwrite(str(dest_img), img)

    # Run inference
    results = model.predict(img, conf=0.15, verbose=False)[0]

    lines = []
    for box in results.boxes:
        cls_id = int(box.cls[0].item())
        # xywhn gives normalized cx, cy, w, h directly for YOLO
        cx, cy, bw, bh = box.xywhn[0].tolist()
        lines.append(f"{cls_id} {cx:.6f} {cy:.6f} {bw:.6f} {bh:.6f}")

    # Write label file directly to disk
    with open(dest_lbl, "w") as f:
        f.write("\n".join(lines))

    if (idx + 1) % 50 == 0 or (idx + 1) == len(image_files):
        print(f"[{idx + 1}/{len(image_files)}] Images labeled and saved!")

print("All labels generated successfully!")
```

```python
!pip install -q reportlab gradio ultralytics opencv-python-headless pillow numpy

import gradio as gr
import cv2
import numpy as np
from PIL import Image
from ultralytics import YOLO
from reportlab.lib.pagesizes import letter
from reportlab.pdfgen import canvas
import datetime
import tempfile
import os

# Load model weights if present
model = None
if os.path.exists("best.pt"):
    try:
        model = YOLO("best.pt")
    except Exception:
        model = None
elif os.path.exists("runs/detect/kandadrishti_model/weights/best.pt"):
    try:
        model = YOLO("runs/detect/kandadrishti_model/weights/best.pt")
    except Exception:
        model = None

def generate_pdf(stats, batch_id="DOCA-2026-SIH"):
    temp_pdf = tempfile.NamedTemporaryFile(delete=False, suffix=".pdf")
    c = canvas.Canvas(temp_pdf.name, pagesize=letter)

    # Header
    c.setFont("Helvetica-Bold", 16)
    c.drawString(50, 750, "KandaDrishti AI - Quality Assessment Certificate")
    c.setFont("Helvetica", 10)
    c.drawString(50, 735, f"Authority: Department of Consumer Affairs (DoCA) | Batch: {batch_id}")
    c.drawString(50, 720, f"Generated On: {datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    c.line(50, 710, 550, 710)

    total = max(stats["Total"], 1)
    grade_a_pct = (stats["Grade_A"] / total) * 100
    grade_b_pct = (stats["Grade_B"] / total) * 100
    urs_pct = (stats["URS"] / total) * 100
    defective_pct = (stats["Defective"] / total) * 100

    c.setFont("Helvetica-Bold", 12)
    c.drawString(50, 680, "Grading Assessment Breakdown (Agmark / NAFED Standards):")

    c.setFont("Helvetica", 11)
    c.drawString(70, 650, f"Total Count Inspected: {stats['Total']} units")
    c.drawString(70, 625, f"Grade A (>45 mm equatorial dia): {stats['Grade_A']} ({grade_a_pct:.1f}%)")
    c.drawString(70, 600, f"Grade B (40 - 45 mm equatorial dia): {stats['Grade_B']} ({grade_b_pct:.1f}%)")
    c.drawString(70, 575, f"URS (<40 mm Under Regular Size): {stats['URS']} ({urs_pct:.1f}%)")
    c.drawString(70, 550, f"Defective (Rotten / Sprouted / Damaged): {stats['Defective']} ({defective_pct:.1f}%)")

    c.line(50, 520, 550, 520)
    c.setFont("Helvetica-Bold", 12)

    status = "ACCEPTED FOR BUFFER STOCK PROCUREMENT" if (defective_pct < 5.0 and grade_a_pct >= 50.0) else "DISPUTED / MANUAL RE-INSPECTION REQUIRED"
    c.drawString(50, 490, f"Procurement Verdict: {status}")
    c.save()
    return temp_pdf.name

def check_defect(crop_bgr):
    """Inspects onion surface for rot (black/dark mold) and sprouting (green shoot)"""
    hsv = cv2.cvtColor(crop_bgr, cv2.COLOR_BGR2HSV)

    # Green detection (sprouted)
    lower_green = np.array([35, 40, 40])
    upper_green = np.array([85, 255, 255])
    green_mask = cv2.inRange(hsv, lower_green, upper_green)
    green_ratio = np.sum(green_mask > 0) / max(crop_bgr.shape[0] * crop_bgr.shape[1], 1)

    # Dark/Black detection (rotten)
    lower_black = np.array([0, 0, 0])
    upper_black = np.array([180, 255, 45])
    black_mask = cv2.inRange(hsv, lower_black, upper_black)
    black_ratio = np.sum(black_mask > 0) / max(crop_bgr.shape[0] * crop_bgr.shape[1], 1)

    if green_ratio > 0.04:
        return "Sprouted"
    elif black_ratio > 0.08:
        return "Rotten"
    return "Sound"

def analyze_onion_tray(image_file):
    if image_file is None:
        return None, "⚠️ Please upload an image or capture a photo.", None

    # Handle image format and color space cleanly
    if isinstance(image_file, np.ndarray):
        pil_img = Image.fromarray(image_file).convert("RGB")
    else:
        pil_img = image_file.convert("RGB")

    img_rgb = np.array(pil_img)
    img_bgr = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2BGR)
    h_img, w_img, _ = img_bgr.shape

    annotated = img_bgr.copy()
    stats = {"Grade_A": 0, "Grade_B": 0, "URS": 0, "Defective": 0, "Total": 0}
    boxes = []

    # 1. Try YOLO Model first with low confidence
    if model is not None:
        try:
            results = model.predict(img_bgr, conf=0.04, iou=0.40, device="cpu", verbose=False)[0]
            for box in results.boxes:
                cls_name = model.names[int(box.cls[0])]
                if "coin" not in cls_name:
                    xyxy = box.xyxy[0].cpu().numpy().astype(int)
                    boxes.append((xyxy[0], xyxy[1], xyxy[2] - xyxy[0], xyxy[3] - xyxy[1]))
        except Exception:
            pass

    # 2. Adaptive Vision Segmentation (Guarantees detection on studio shots or trays)
    if len(boxes) == 0:
        gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
        blur = cv2.GaussianBlur(gray, (5, 5), 0)
        _, thresh = cv2.threshold(blur, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

        kernel = np.ones((5, 5), np.uint8)
        opening = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel, iterations=2)

        contours, _ = cv2.findContours(opening, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
        for cnt in contours:
            x, y, w, h = cv2.boundingRect(cnt)
            if w > 45 and h > 45 and (w * h) < (w_img * h_img * 0.85):
                boxes.append((x, y, w, h))

    # 3. Millimeter scaling
    px_to_mm = 0.16  # 1 px ≈ 0.16 mm at standard camera distance

    # 4. Classify each detected sample
    for (x, y, w, h) in boxes:
        crop = img_bgr[max(0, y):min(h_img, y + h), max(0, x):min(w_img, x + w)]
        defect_type = check_defect(crop) if crop.size > 0 else "Sound"

        diameter_mm = round(max(w, h) * px_to_mm, 1)

        if defect_type != "Sound":
            category = f"Defective ({defect_type})"
            stats["Defective"] += 1
            color = (0, 0, 255)  # Red
        else:
            if diameter_mm >= 45.0:
                category = "Grade A"
                stats["Grade_A"] += 1
                color = (0, 200, 0)  # Green
            elif diameter_mm >= 40.0:
                category = "Grade B"
                stats["Grade_B"] += 1
                color = (0, 255, 255)  # Yellow
            else:
                category = "URS"
                stats["URS"] += 1
                color = (255, 140, 0)  # Orange

        stats["Total"] += 1

        cv2.rectangle(annotated, (x, y), (x + w, y + h), color, 3)
        cv2.putText(annotated, f"{category} | {diameter_mm}mm", (x, max(y - 8, 20)),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.55, color, 2)

    total = max(stats["Total"], 1)
    summary_md = f"""
### 🧅 Batch Analysis Summary
* **Total Samples Detected:** {stats['Total']} units
* **Grade A (>45 mm):** {stats['Grade_A']} ({(stats['Grade_A']/total)*100:.1f}%)
* **Grade B (40-45 mm):** {stats['Grade_B']} ({(stats['Grade_B']/total)*100:.1f}%)
* **URS (<40 mm Under Regular Size):** {stats['URS']} ({(stats['URS']/total)*100:.1f}%)
* **Defective Samples:** {stats['Defective']} ({(stats['Defective']/total)*100:.1f}%)
"""

    annotated_rgb = cv2.cvtColor(annotated, cv2.COLOR_BGR2RGB)
    pdf_path = generate_pdf(stats)
    return annotated_rgb, summary_md, pdf_path

# Gradio Interface
with gr.Blocks(title="KandaDrishti AI", theme=gr.themes.Soft()) as demo:
    gr.Markdown("# 🧅 KandaDrishti AI: Automated Onion Grading System")
    gr.Markdown("Department of Consumer Affairs (DoCA) | Smart India Hackathon Prototype")

    with gr.Tabs():
        with gr.TabItem("📁 Option 1: File / Gallery Upload"):
            file_in = gr.Image(type="pil", label="Upload from Gallery or Camera", sources=["upload"])
            file_btn = gr.Button("🔍 Analyze Image", variant="primary")

        with gr.TabItem("📸 Option 2: Live Camera Capture"):
            cam_in = gr.Image(type="pil", label="Capture via Camera", sources=["webcam"])
            cam_btn = gr.Button("🔍 Analyze Live Capture", variant="primary")

    out_img = gr.Image(label="AI Detection, Diameter Sizing & Quality Grading Output")
    summary = gr.Markdown()
    pdf_download = gr.File(label="📥 Download Official Quality Certificate (PDF)")

    file_btn.click(fn=analyze_onion_tray, inputs=file_in, outputs=[out_img, summary, pdf_download])
    cam_btn.click(fn=analyze_onion_tray, inputs=cam_in, outputs=[out_img, summary, pdf_download])

    demo.launch(share=True)
```
